## JUEGO DE ADIVINANZAS

#### Escribe un programa que coja un número entero del 1 al 100, y los jugadores tienen que tratar de adivinar el número.

#### Las reglas son las siguientes:

1. Si un jugador escoje un número menor que 1 o mayor que 100, el programa dice "FUERA DE LÍMITES"
2. En el primer turno del jugador, si está en un rango de 10 del número dice "¡CALIENTE!". Si está fuera de ese rango dice "¡FRIO!"
3. En todas las rondas siguientes, si el jugador se acerca más que la vez anterior, dice "¡MAS CALIENTE!". Si está más lejos que la anterior vez dice "¡MAS FRIO!"
4. Cuando el jugador acierte el número, les dirá que han acertado y cuántos intentos les ha llevado





##### PISTAS

5. Modifica ese bucle para que compare nuestro número con el intento. Si se acierta usa break, si no puedes usar continue y tienes que seguir preguntando al jugador por nuevos intentos

```python
import random
from random import randint
numero=random.randint(1,100)
print(numero)
```

```python
print("BIENVENIDO A: DA CONMIGO")
print("Así funciona:")
print("Voy a pensar un número del 1 al 100 y tú tendrás que adivinarlo.""\n""Si tu intento de adivinarlo es de una diferencia >10 con mi número, el juego te devolverá: FRÍO.""\n""De froma contraria, si la diferencia es <10, te devolverá CALIENTE.""\n""A medida que vayas generando más intentos, si la diferencia de los nuevos es mayor que la de tu intento anterior, el juego devolverá MÁS FRÍO.""\n""De froma contraria, devovlerá MÁS CALIENTE.""\n""Cuando des con mi número, el juego devolverá la cantidad de intentos que has necesitado y podrás competir con tus amigos para ver quien necesita menos intentos.""\n""Ten en cuenta que los números se generan al azar cada vez que juguéis.")
lista_intentos=[]
while True:
    intento=input("Introduce un número entre el 1 y el 100:")
    intento=int(intento)
    intento
    if intento > 100 or intento < 1:
        print("FUERA DE LÍMITES")
    else:
        if len(lista_intentos)==0:
            if intento == numero:
                print("HAS ACERTADO EN EL PRIMER INTENTO")
                break
            elif abs(intento-numero) <=10:
                lista_intentos.append(intento)
                print("CALIENTE")
            else:
                lista_intentos.append(intento)
                print("FRÍO")
        else:
            if intento == numero:
                print(f"Acertaste con {len(lista_intentos)} intentos")
                break
            elif abs(numero-lista_intentos[-1])<abs(numero-intento):
                lista_intentos.append(intento)
                print("MÁS FRÍO")
            else:
                lista_intentos.append(intento)
                print("MÁS CALIENTE")
```
